**DOUBLY LINKED LIST FROM ARRAY**

gfg question

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None
        self.prev = None
 
class Solution:
    def createDLL(self,arr):
        if not arr:
            return None
        
        self.head = Node(arr[0])
        current = self.head

        for i in range(1, len(arr)):
            new_node = Node(arr[i])
            current.next = new_node 
            new_node.prev = current
            current = new_node          

        return self.head
```

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None
        self.prev = None
class Solution:
    def insertAtPos(self,head,p,x):
        new_node = Node(x)
        
        curr = head 
        count = 0
        
        while curr is not None and count < p:
            curr = curr.next
            count += 1
        if curr is None:
            return head 
        new_node.prev = curr.prev
        new_node.next = curr
        
        if curr.next:
            curr.prev.next = new_node
            
        curr.next = new_node
        return head
```
